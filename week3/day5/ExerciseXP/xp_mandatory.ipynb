> # Designing an Evaluation Strategy for Different ML Models

## 1. Supervised Learning (e.g., Classification via Logistic Regression for Loan Default)

**Evaluation Strategy:**
- **Cross-validation** — 5-fold stratified CV on the training set to tune hyperparameters and estimate variance.
- **Metrics** — accuracy (only as sanity check), **precision, recall, F1-score** (imbalanced classes)
- **Threshold tuning** — lower the decision threshold (e.g., 0.2–0.3) based on business cost of FN vs FP.

---

## 2. Unsupervised Learning (e.g., K-Means Clustering for Customer Segmentation)

**Evaluation Strategy:**
- **Silhouette score** — measures how similar a point is to its own cluster vs. other clusters (range -1 to 1, higher = better).
- **Elbow method** — plot inertia (sum of squared distances) vs. #clusters; pick the "elbow" where gain drops off.

**Challenges/Limitations:**
- No ground truth → evaluation is inherently subjective / heuristic.
- K-means assumes spherical, equal-sized clusters; sensitive to scaling and initialization.
- Results depend heavily on feature selection and preprocessing.

---

## 3. Reinforcement Learning (e.g., Q-Learning for Robot Navigation)

**Evaluation Strategy:**
- **Cumulative reward (return)** — total discounted reward over an episode; compare against a baseline/random policy.
- **Convergence** — plot episode reward over time; check if it stabilizes (plateaus) — indicates the policy has learned.
- **Exploration vs. exploitation balance** — track exploration rate (ε) decay and whether the agent re-visits poorly-known states vs. exploiting known good actions.

**Challenges/Limitations:**
- **Non-stationarity** — the environment changes as the agent learns (data distribution shifts) → tricky to validate.
- **No clean train/test split** — evaluation happens in the live environment, so it's expensive/risky.
- **Reward shaping** — poorly designed reward functions can lead to unexpected policies.
- High variance between runs → need multiple seeds and statistical comparison.

---

## 4. Cross-Category Challenges Summary

| Category | Key Challenge |
|----------|---------------|
| Supervised | Imbalance + overfitting + interpretability |
| Unsupervised | No ground truth → subjective validation |
| Reinforcement | No test set → online evaluation, non-stationarity, seed variance |
