- Create a section in your document that clearly differentiates between Supervised, Unsupervised, and Reinforcement Learning.
For each type, provide a brief definition and an example scenario where that type of machine learning is typically applied.

### **Answer**:

**Examples**:
- 1. Capital Market: predicting stock market movements
- 2. Banking/Payments: detection of fraudelent behavior
- 3. HealthCare: disease diagnostics

**Supervise Learning**: clearly labeled outcomes to train model on and evaluate live performance. Goal: predict or classify

**Unsupervised Learning**: no labeled outcomes; the model tries to idetify patterns in the data and typically group/cluster based on those patterns. Hence, most prominent application is clustering. Goal: to cluster or associate

**Reinforcement Learning**: no labaled outcomes, but a clear reward function providing feedback to the model to act on. There is no actual training data set - training is done in 'live' mode going forward based of fresh data. Typically used in action/execution oriented tasks e.g. self-driving car decisions.
Goal: for model to make decisions/act and derive consequences from these for following decisions/actions





- Describe the process of developing a machine learning model. Focus on three main stages: Feature Selection, Model Selection, and Model Evaluation.

### **Answer**:

**Three main stages of ML development**

0.   Data collection, EDA, cleaning

1.   Feature Selection:  preliminary correlation and group averages checks to identify usable and relevant input data. Then dimensionality reduction (PCA) to make the future model smaller and less computation heavy.

2.   Model Selection: depending on the learning type, choose an aproopriate model

3.   Model Evaluation:
- Error evautation/Loss function: depending on the objective, selection of an appropriate error metric - if it's about classification, then accuracy errors (precision, recall etc.), if value predictions then rather various regressions and mean errors, and if
- Splitting data into test/validatation/training data sets to minimize chances of overfitting and get a robust model. Important to avoid biases -> stratified sampling, and oftentimes a good approach in each training run to keep a different fraction of test data for validation to achieve robust model parameterizations
