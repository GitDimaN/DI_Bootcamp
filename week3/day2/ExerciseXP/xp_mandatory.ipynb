**Exercise 1 : Matrix Operations**

**Instructions:**
In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

**Determinant**: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it you can watch this video.

**Inverse of a Matrix**: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.
Create a 3x3 matrix and perform the following operations:

Calculate the determinant.
Find the inverse of the matrix.

```python
rng = np.random.default_rng(seed=42)
matrix_3_3 = rng.integers(low=1, high=50, size=(3, 3))
matrix_3_3
```

```python
# Determinant (scalar)
det = np.linalg.det(matrix_3_3)

# Inverse
matrix_3_3_inv = np.linalg.inv(matrix_3_3)

print("Determinant:", det)
print("Inverse:\n", matrix_3_3_inv)

# Quick verification: A @ A_inv ≈ Identity
print("A @ A_inv:\n", np.round(matrix_3_3 @ matrix_3_3_inv, 10))
```

Images are represented by pixel data that can be stored in a numpy array - example of importing image data into an numpy array and converting it to grayscale below:

```python
# Images are represented as an array of pixel data-lists - 3 RGB colors and opacity.
# Gray shade is simply a color where Red, Green and Blue have equal intensity
# Grayscale pixels can actually either be a single value (intensity 0-255) or represented as [Y, Y, Y, 255] in RGBA.

# So, we creating a 5x5 np array for 5x5 grayscale "directions" first, via 1 value for 1 pixel

gray_values = np.array([
    [10, 50, 100, 150, 200],
    [30, 70, 120, 170, 220],
    [50, 90, 140, 190, 240],
    [70, 110, 160, 210, 250],
    [90, 130, 180, 230, 255]
], dtype=np.uint8)
print("5x5 Grayscale values:\n", gray_values)

# Alternatively, in 3D RGBA representation [Y, Y, Y, A] by stacking each gray-schade 3 times to get RGBs that are all equal for that pixel
gray_rgba = np.stack([gray_values, gray_values, gray_values, np.full((5, 5), 255, dtype=np.uint8)], axis=-1)
print("5x5 Grayscale RGBA representation:\n", gray_rgba)

# Now we can display the image using matplotlib. Both gray_values or gray_rgba can be used:
plt.imshow(gray_values, cmap='gray', vmin=0, vmax=255)
plt.title("5x5 Grayscale Image")
plt.colorbar()
plt.show()
```

**Plotting a True 5x5 Pixel Image**

To display the image exactly as a $5 \times 5$ pixels on the screen, we can configure Matplotlib to use a specific figure size in inches and DPI (Dots per inch), such that the count of Dots results in 5 wide and high. And disable all margins and axes.

```python
# Determine the figure size in inches for exactly 5x5 screen pixels
# We set DPI to 100, so a 5x5 pixel image is 0.05 x 0.05 inches
dpi = 100
fig_size = 5 / dpi

fig = plt.figure(figsize=(fig_size, fig_size), dpi=dpi)
ax = fig.add_axes([0, 0, 1, 1]) # Fill the entire figure frame (no margins)

# Display the array with zero interpolation
ax.imshow(gray_values, cmap='gray', vmin=0, vmax=255, interpolation='none')

# Turn off axes, labels, and ticks to prevent distortion
ax.axis('off')

plt.show()
```

Now we can also save it directly as a PNG using Pillow:

```python
# Save the raw 5x5 image directly to a physical PNG file
from PIL import Image

# Create the image from our uint8 array
img_to_save = Image.fromarray(gray_values)

# Save to disk
output_path = '/content/unscaled_5x5_image.png'
img_to_save.save(output_path, format='PNG')
print(f"Success! Your unscaled 5x5 image has been saved to: {output_path}")
```

**Exercise 6 : Basic Hypothesis Testing**

**Instructions:**

Create a sample dataset to test the effectiveness of a new training program on employee productivity.

Given a dataset below, formulate a simple hypothesis and test it using basic statistical functions in NumPy.

```python
# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness
# and test it using basic statistical functions in NumPy.

# Hypothesis H0 is that there is no increased productivty after the training program.
# I.e. the sample averages of productivity scores before/after aren't significantly different
# we use alpha =0.05

# via scipy
import scipy.stats as stats

# Perform a two-sample t-test
t_test, pval = stats.ttest_ind(productivity_before, productivity_after)

# Print the results
print("t-statistic:", t_test)
print("p-value:", pval)

# Interpreting the result
alpha = 0.05  # Significance level
if pval < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

Since p_value is >0.05, we can't the reject the Hypothesis H0.

**Exercise 8 : Time Series Data Manipulation**

**Instructions:**
Generate time series data for the year 2023. Demonstrate slicing for the following intervals:

- January to March
- April to June
- July to September

- October to December


Generate a time series data for a specific period and demonstrate how to slice this data for different intervals.

```python
import numpy as np

# Generate time series data for the entire year of 2023
dates_arr = np.arange('2023-01-01', '2024-01-01', dtype='datetime64[D]')

# Slice the array into the four quarters by converting comparison strings to np.datetime64
q1_2023 = dates_arr[(dates_arr >= np.datetime64('2023-01-01')) & (dates_arr <= np.datetime64('2023-03-31'))]
q2_2023 = dates_arr[(dates_arr >= np.datetime64('2023-04-01')) & (dates_arr <= np.datetime64('2023-06-30'))]
q3_2023 = dates_arr[(dates_arr >= np.datetime64('2023-07-01')) & (dates_arr <= np.datetime64('2023-09-30'))]
q4_2023 = dates_arr[(dates_arr >= np.datetime64('2023-10-01')) & (dates_arr <= np.datetime64('2023-12-31'))]

# Display the slices
print(f"January to March (Q1): {q1_2023[0]} to {q1_2023[-1]} (Total days: {len(q1_2023)})")
print(f"April to June (Q2): {q2_2023[0]} to {q2_2023[-1]} (Total days: {len(q2_2023)})")
print(f"July to September (Q3): {q3_2023[0]} to {q3_2023[-1]} (Total days: {len(q3_2023)})")
print(f"October to December (Q4): {q4_2023[0]} to {q4_2023[-1]} (Total days: {len(q4_2023)})")
```
