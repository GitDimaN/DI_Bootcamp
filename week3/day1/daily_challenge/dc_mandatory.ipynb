```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Tasks:

**1. Data Preparation:**

**Hint 1:** Use np.random.uniform(low, high, size) to generate the temperature data.

**Hint 2:** Create a DataFrame using pd.DataFrame(data, index, columns) with appropriate index and columns.

Use NumPy to generate a synthetic dataset representing average monthly temperatures (in degrees Celsius) for 12 months across 10 different cities. The temperatures should range from -5 to 35 degrees.

Convert this NumPy array into a Pandas DataFrame, adding city names as index and months as columns.

```python
# city names
city_arr = np.array(['Tokyo', 'Delhi', 'Shanghai', 'Mexico City','Cairo', 'Mumbai', 'Beijing',
    'New York', 'Karachi', 'Buenos Aires'], dtype='str')
city_arr
```

```python
# random generaiton of 10x12 monthly avg temperatures for 10 cities:
temp_arr = np.random.uniform(-5,35,size=(10,12))
temp_arr
```

```python
# converting into a data framce:
df = pd.DataFrame(temp_arr, index=city_arr, columns=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
df
```

**2. Data Analysis:**

**Hint 1:** Calculate the annual average temperature using DataFrame.mean(axis).

**Hint 2:** Find the city with the highest and lowest average temperature using idxmax() and idxmin() methods.

Calculate the annual average temperature for each city.

Identify the city with the highest and lowest average temperature for the year.

```python
df['city_avg_temp'] = df.mean(axis=1)
df
```

```python
# finding max and min locations among the city average temps:
print(f"The city with the highest average annual temperature is: {df['city_avg_temp'].idxmax()}")
print(f"The city with the lowest average annual temperature is: {df['city_avg_temp'].idxmin()}")
```

**3. Data Visualization**

```python
#Expanding the overall figure width to handle a wide legend inside the plot at the bottom
plt.figure(figsize=(13, 6))
plt.plot(df.iloc[:,0:-1].T, linewidth=1.5, marker='o', markersize=4)
plt.ylim(bottom = -15) # Changed to -15 to make extra room for the larger legend
plt.xticks(rotation = 45)
plt.title("Monthly Temperatures Across Cities")
plt.ylabel("Temperature (°C)")
plt.xlabel("Months")

# Legend labels that will include the city name and its average temperature in brackets
# best way: legend_labels = [f"{city} ({df.loc[city, 'city_avg_temp']:.1f}°C)" for city in df.index]
legend_labels = [f"{df.index[i]} ({temp:.1f}°C)" for i,temp in enumerate(df['city_avg_temp'])]

# Advanced legend styling - placed inside the plot at 'lower center'
plt.legend(
    legend_labels,          # Custom labels with average temperatures
    loc='lower center',     # Inside the plot
    ncol=5,                 # Keep 5 columns
    frameon=True,          # Keep the border box visible
    facecolor='white',     # Background color of the legend box
    edgecolor='gray',      # Border line color
    shadow=True,           # Give the legend box a drop shadow
    fontsize=11            # Increased font size to make the legend larger
)

plt.tight_layout()
plt.show()
```
